```python
# merging dataframes
# inner merge >> merge based on common elements
# right merge >> merge based on elements from the right
# left merge >> merge based on elements from the left
# outter merge >> all elements of both the dataframes
# cross merge >> every permutation & combination of elements are there
```

```python
df1 = pd.DataFrame({
    'key 1' : [1, 2, 4, 5, 6],
    'key 2' : [4, 5, 6, 7, 8],
    'key 3' : [3, 4, 5, 6, 6]
})
df1
```

```python
df2 = pd.DataFrame({
    'key 1' : [1, 2, 45, 6, 67],
    'key 4' : [56, 5, 6, 7, 8],
    'key 5' : [3, 56, 5, 6, 6]
})
df2
```

```python
# merging df1 and df2
# inner merge
pd.merge(df1, df2, how = "inner")
```

```python
# left >> all elements from the left dataframe
pd.merge(df1, df2, how = "left")
```

```python
# right >> all elements from the right dataframe
pd.merge(df1, df2, how = "right")
```

```python
# outer >> all the common elements of the common column
print(f"df1:\n{df1}\n{"***"*20}\ndf2:\n{df2}\n{"***"*20}")

print("Outer merge: df1, df2")
pd.merge(df1, df2, how = 'outer')
```

```python
# cross join >> permutation of dataframes
pd.merge(df1, df2, how = "cross")
```

```python
# using specific columns to merge dataframes
pd.merge(df1, df2, how = 'left', left_on = "key 2", right_on = "key 4")
```

```python
# join >> join works based on index
# merge >> merge works based on columns
df1 = pd.DataFrame({
    'key 1' : [1, 2, 4, 5, 6],
    'key 2' : [4, 5, 6, 7, 8],
    'key 3' : [3, 4, 5, 6, 6]
}, index = ['a', 'b', 'c', 'd', 'e'])
df1
```

```python
df2 = pd.DataFrame({
    'key 6' : [1, 2, 45, 6, 67],
    'key 4' : [56, 5, 6, 7, 8],
    'key 5' : [3, 56, 5, 6, 6]
}, index = ['a', 'b', 'h', 'i', 'j'])
df2
```

```python
# joining columns of dataframes based on common indexes
df1.join(df2, how = 'inner')
```

```python
# joining left
df1.join(df2, how = 'left')
```

```python
# joining right
df1.join(df2, how = 'right')
```

```python
df1.join(df2, how = 'outer')
```

```python
df1.join(df2, how='cross')
```

```python
# converting fare from euro to inr
# assuming euro 1 = inr 90
# obj.apply() >> to apply a function on all the elements/values of a column
df['Fare_inr'] = df['Fare'].apply(lambda x: x*90)
```

```python
# creating a column for storing length of the name
df["Name_len"] = df['Name'].apply(lambda x: len(x))
```

```python
# applying user defined function
def conversion(x):
    return x*90
```

```python
# applying the above defined funtion to "Fare" column from original dataframe
df["Fare1"] = df["Fare"].apply(conversion)
```

```python
df.head()
```

```python
# create a column which shows the status of fare: cheap, medium, high
def create_flag(x):
    if x < 10:
        return "cheap"
    if x >= 10 and x < 20:
        return "medium"
    else:
        return "high"
```

```python
df["Fare_flag"] = df["Fare"].apply(create_flag)
```
