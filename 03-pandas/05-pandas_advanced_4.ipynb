```python
# creating new dataframe df2
df2 = pd.DataFrame({'a': [1, 2, 3, 4, 5, 6, 7, 8, 9]})
df2
```

```python
# average on rolling basis: example (average of first two numbers)
df2.rolling(window = 2).mean()
# window >> average of how many numbers
```

```python
# average value of 3 values on rolling basis
df3 = df2.rolling(window = 3).mean()
# NaN >> there are no average value available for 1st 3 values hence we get NaN
df3
```

```python
# dropping None/NaN values from the dataframe: df3
df3.dropna(inplace = True)
df3
```

```python
# typecasting >> converting data types
df3['a_int'] = df3['a'].astype(int)
```

```python
df3
```

```python
# rolling sum >> find sum of values on rolling basis for window size 2
df2.rolling(window = 2).sum()
```

```python
# date time concept (in pandas)
df = pd.DataFrame({'date': ['2024-03-08', '2024-03-09', '2024-03-10']})
df
```

```python
df.info()
```

```python
df.dtypes
```

```python
# the data has a dtype of object: but we want it in date-time
# so we'll convert manually
df['updated_date'] = pd.to_datetime(df['date'])
df
```

```python
# check dtype
df.dtypes
```

```python
# using 'datetime' make it easy to categorise dd,mm,yyyy
# getting year from datetime
df['updated_date'].dt.year
```

```python
df['year'] = df['updated_date'].dt.year
```

```python
df
```

```python
# getting day or month from datetime
df['month'] = df['updated_date'].dt.month
print(df)

print("***"*15)

df['day'] = df['updated_date'].dt.day
print(df)
```

```python
# date range >> generate data of dtype: datetime
pd.date_range(start = '2024-07-10', end = '2025-07-10', freq = 'D')
# example: generating daily data from one day to the same day next year
```

```python
# generating data on hourly basis (same range)
pd.date_range(start = '2024-07-10', end = '2025-07-10', freq = 'H')
```

```python
# for future use: freq = 'H' has been replaced with freq = 'h'
pd.date_range(start = '2024-07-10', end = '2025-07-10', freq = 'h')
```

```python
# get current time
pd.Timestamp.now()
```

```python
# to check the execution time of a program/code
start = pd.Timestamp.now()
for i in range(1000):
    pass
end = pd.Timestamp.now()
print(f"Time taken for code execution: {end - start}")
```

```python
# another way: importing a module: time
import time
start = time.time()
for i in range(1000):
    pass
end = time.time()
print(f"Execution time: {end - start} seconds")
```

```python
# even another way: best way so far (also use is industry)
import time
start = time.perf_counter()
for i in range(1000):
    pass
end = time.perf_counter()
print(f"Execution time: {end - start} seconds")
```
