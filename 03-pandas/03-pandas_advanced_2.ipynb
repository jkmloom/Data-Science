```python
import pandas as pd

# creating new dataframe with new (custom) data: for understanding
data1 = {
    'A' : [1, 2, None, 4, 5, None, 7, 8, 9, 10],
    'B' : [None, 11, 12, 13, None, 15, 16, None, 18, 19]
}
df2 = pd.DataFrame(data1)
```

```python
# review new dataframe
df2
```

```python
# replacing missing values with the mean value
df2['A'].mean()
# calculated column 'A' mean
```

```python
df2['A'].fillna(df2['A'].mean()) # replaced/imputed the null value swith mean
```

```python
# replacing/imputing the missing values with median
df2["A"].fillna(df2['A'].median())
```

```python
# replace/impute all the null values with some value (mean)
df2.fillna(round(df2.mean(), 2))
```

```python
# another way to replace/impute the null data
df2.fillna(method = "ffill")
# ffill >> stands for 'forward fill'
```

```python
# updated way for using "ffill"
# ffill >> forward fill >> start seeing daat from the front and fill with the next observed value
df2.ffill()
```

```python
# bfill >> backward fill >> start seeing data from the back and fill with next observed value
df2.bfill()
```

```python
# check for duplicate row
df2.duplicated().sum()
```

**Working of obj.groupby()**  


```python
# we can also perform other operations/functions on obj.groupby()
df.groupby('Survived').describe()
```

```python
# Now suppose we've to work with 'Survived' and 'Fare' columns
df[['Survived', 'Fare']]
```

```python
import numpy as np
df.groupby(['Survived'])['Fare'].agg(['min', 'max', 'mean', 'median', 'count', np.std, 'var'])
```

```python
# groupby based on two columns
# Q. How many pasenger survived/not survived for each sex and pclass
df.groupby(['Sex', 'Pclass'])['Survived'].sum()
```

```python
# convert above study into a separte dataframe
df.groupby(['Sex', 'Pclass'])['Survived'].sum().to_frame()
```

```python
# reset the index for this dataframe
df.groupby(['Sex', 'Pclass'])['Survived'].sum().to_frame().reset_index()
```
