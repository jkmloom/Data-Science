```python
# age column has string value, but we need it to be int

# 'coerce' turns non-numbers into NaN (Not a Number)
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
# Since NaN is a float, you'll need to fill them before converting to int
df['Age'] = df['Age'].fillna(0).astype(int)
```
