```python
import numpy as np
```

```python
# return an array of zeros for a given dimention and size
np.zeros(3)
```

```python
# dtype is used to define the data type (default=float)
np.zeros((3, 3), dtype=int)
```

```python
array_var = np.zeros((3, 4), dtype=int)
print(f"Array:\n{array_var}\nDimentions: {array_var.ndim}D")
```

```python
# return an array filled with ones
np.ones((3, 4), dtype=int)
# default dtype=float
```

```python
np.ones((2, 2))
```

```python
# creating an array of twos
array_var = np.ones((3, 3), dtype=int)*2
print(f"Array:\n{array_var}")
```

```python
# another method
np.array([[2, 2 ,2], [2, 2, 2], [2, 2, 2]])
```

```python
# creating an 3D array
print(f"1D Array: {np.zeros(2)}")
print(f"{'---'*12}\n2D Array:\n{np.zeros((2, 2))}")
print(f"{'---'*12}\n3D Array:\n{np.zeros((3, 2, 2))}") 
```

```python
array_3d = np.zeros((3, 3, 4), dtype=int)
print(f"Array:\n{array_3d}")
print(f"Dimantions: {array_3d.ndim}D")
```

```python
# creating 4D array
array_4d = np.ones((1, 1, 3, 4), dtype=int)
print(f"Array:\n{array_4d}")
print(f"Dimentions: {array_4d.ndim}D")
```

```python
# creating 5D array
array_5d = np.ones((1, 1, 1, 3, 4), dtype=int)*5
print(f"Array:\n{array_5d}")
print(f"Dimentions: {array_5d.ndim}D")
```

```python
arr = np.zeros((3, 4), dtype=int)
print(arr + 5)
```
