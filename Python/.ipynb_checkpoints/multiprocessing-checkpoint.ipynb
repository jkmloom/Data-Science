```python
import time

start = time.perf_counter()

def test_func():
    print("Sleeping for a second..")
    time.sleep(1)
    print("Done sleeping!")

test_func()
test_func()

end = time.perf_counter()
print(f"Time taken by this program to execute: {round(end-start, 2)} secs")
```

```python
# achieving above result using multiprocessing 
import multiprocessing
# multiprocessing >> using multiple processors/cores (multiprocessing is parallel)
import time

start = time.perf_counter()
def test_func():
    print("Sleeping for a second")
    time.sleep(1)
    print("Done sleeping!")

p1 = multiprocessing.Process(target = test_func())
p2 = multiprocessing.Process(target = test_func())

p1.start()
p2.start()

p1.join()
p2.join()

end = time.perf_counter()
print(f"The program executed in {round(end-start, 2)} secs")
```

```python
# generealized version of the above code
import multiprocessing
import time

start = time.perf_counter()
def test_func():
    print("Sleeping for a second")
    time.sleep(1)
    print("Done sleeping!")

processes = []
for i in range(3):
    p = multiprocessing.Process(target = test_func())
    p.start()
    processes.append(p)

for process in processes:
    process.join()

end = time.perf_counter()
print(f"The program executed in {round(end-start, 2)} secs")
```

```python
# multiprocessing should be used in cases where computation matters
# use case 1:
import multiprocessing
import time

start = time.perf_counter()

def square(index, value):
    value[index] = value[index] ** 2

arr = multiprocessing.Array('i', [1, 2, 5, 3, 400])
processes = []

for i in range(len(arr)):
    p = multiprocessing.Process(target = square(i, arr))
    p.start()
    processes.append(i)

# for process in processes:
#     process.join()

print(f"{list(arr)}")

end = time.perf_counter()
print(f"Program executed in {round(end-start, 2)}")
```

```python
# using multiprocessing.pool()
import multiprocessing
import time
import sys

def square(num):
    result = num * num
    print(f"The square of {num} is {result}", flush=True)

if __name__ == "__main__":
    start = time.perf_counter()

    numbers = [1, 2, 3, 4, 6000]
    
    with multiprocessing.Pool() as pool:
        print("Starting Pool.map")
        pool.map(square, numbers)
        print("Finished Pool.map")

    end = time.perf_counter()
    print(f"The program executed in {round(end-start, 2)} secs")
```

```python
import time
from multiprocessing.dummy import Pool  # uses threads

def square(num):
    result = num * num
    print(f"The square of {num} is {result}", flush=True)

start = time.perf_counter()

numbers = [1, 2, 3, 4, 6000]

with Pool() as pool:
    print("Starting Pool.map")
    pool.map(square, numbers)
    print("Finished Pool.map")

end = time.perf_counter()
print(f"The program executed in {round(end-start, 2)} secs")
```

> The order of the result is not fixed showing that multiple processors/cores were used for executing this program  
- Also proving **Parallel execution of the program**

```python
# Use case 2:
# say you want to get admission in a school for which you're required to make enrollment request
# enrollment and registration can be done parallely (using multiprocessing)
```

```python
import multiprocessing
import time

def enroll_students(student_queue):
    for student in ["Rahul", "Rohit", "Aman", "Ajay"]:
        student_queue.put(f"Enroll request for: {student}")

def register_students(students_queue):
    while True:
        enrollment_request = students_queue.get()
        if enrollment_request is None:
            print("Enrollment request list is empty.")
            break
        print(f"Register the enrollment request: {enrollment_request}")

if True:
    student_queue = multiprocessing.Queue() # multiprocessing has the Queue data structure
    enrollment_process = multiprocessing.Process(target = enroll_students(), args = (student_queue,))
    registration_process = multiprocessing.Process(target = register_students, args = (student_queue,))
    
    enrollment_process.start()
    registration_process.start()

    enrollment_process.join()
    registration_process.join()
```
