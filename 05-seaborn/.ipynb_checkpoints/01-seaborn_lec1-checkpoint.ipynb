# numerical data
## sns.relplot() >> scatter plot
plotting categorical and numerical data  
study relationship between two variabls  
relplot (analogy) relationship plot

```python
df.head(2)
```

```python
# how tip is varying with respect to total bill >> scatter plot
# relplot >> scatter plot (default)
sns.relplot(x='total_bill', y='tip', data=df)
```

```python
# adding hue
sns.relplot(x='total_bill', y='tip', data=df, hue='smoker')
```

in general terms >> as total_bill increases, tip increases  
the higest tip is by a smoker

```python
sns.relplot(x='total_bill', y='tip', data=df, hue='size')
# hue >> used for different color sets
```

```python
sns.relplot(x='total_bill', y='tip', data=df, style='size')
# style >> used to represent a column's values
```

```python
# using both >> style and hue
sns.relplot(x='total_bill', y='tip', data=df, style='smoker', hue='time')
```

## sns.relplot() >> line plot
used usually for data with a variable defining time  
used to study the trend in data  
sequence is generally involved

```python
time = np.arange(0, 500)
value = np.random.randn(500).cumsum()
df = pd.DataFrame({'time':time, 'value':value})
```

```python
df
```

```python
# creating line plot (seaborn)
sns.relplot(x='time', y='value', kind='line', data=df, sort=True)
# kind >> (default=scatter) our value :: line
```

```python
# relplot :: line chart >> used on real dataset
sns.get_dataset_names()
```

```python
fmri = sns.load_dataset('fmri')
fmri.head(3)
```

```python
sns.relplot(x='timepoint', y='signal', kind='line', data=fmri)
```

```python
sns.relplot(x='timepoint', y='signal', kind='line', data=fmri, estimator=None)
```

```python
sns.relplot(x='timepoint', y='signal', kind='line', data=fmri, hue='event', style='region')
```

```python
# remove dashes in graph + adds marker to graph
sns.relplot(x='timepoint', y='signal', kind='line', data=fmri, hue='event', style='region', dashes=False, markers=True)
```

# caterogical data
## sns.catplot()

```python
tips.info()
```

```python
tips.dtypes
```

```python
# plot between day and total_bill
sns.catplot(x='day', y='total_bill', data=tips)
```

**insights** >> more people arrive on weekends compared to rest of the days

```python
# making the plot clearer
# in sns.catplot(), few data-points are overlapping >> solving it here
sns.catplot(x='day', y='total_bill', data=tips, kind='swarm')
# use swarm (parameter)
```

```python
# boxplot >> study the outliers
sns.catplot(x='day', y='total_bill', data=tips, kind='box')
```

```python
# violin plot
sns.catplot(x='day', y='total_bill', data=tips, kind='violin')
```

```python
# boxen plot
sns.catplot(x='day', y='total_bill', data=tips, kind='boxen')
```

```python
tips.head()
```

```python
sns.barplot(x='day', y='total_bill', data=tips)
# study >> total_bill (aggerated) varies with different data
```

```python
sns.jointplot(x=tips.total_bill, y=tips.tip)
# darw a 2 variable plot with distribution >> bivariate plot
```

```python
# studying the relationship with a line
sns.jointplot(x=tips.total_bill, y=tips.tip, kind='reg')
# kind='reg' >> regeration line
```

```python
sns.relplot(x=tips.total_bill, y=tips.tip, hue=tips.smoker)
```

```python
sns.jointplot(x=tips.total_bill, y=tips.tip, kind='hex')
```

```python
sns.jointplot(x=tips.total_bill, y=tips.tip, kind='kde')
# kde >> kernel density estimation
```

- the above plotting is knows as **contour plot**
- example: genrally seen for studying earthquik data or cyclone data
