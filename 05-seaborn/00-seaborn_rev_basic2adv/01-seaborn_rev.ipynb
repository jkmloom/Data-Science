```python
import seaborn as sns
%matplotlib inline
```

```python
# loading a dataset (tips) provided by seaborn
tips = sns.load_dataset('tips')
print(tips)
```

```python
print(tips.dtypes)
```

### **univariant distribution**

```python
# visualizing a distribution
print(sns.displot(tips['total_bill']))
# output has -> histograma and kde (kernel density estimation)
```

```python
sns.displot(tips['total_bill'], kde=False, bins=40) # removed the kde part
# bins='int' >> used for customizing the graph defination for a data
```

### **bivariant distribution**

```python
# combining 2 different distplot() for bivariant data
sns.jointplot(x='total_bill', y='tip', data=tips)
# simply >> combining 2 distplto()s
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='hex')
# kind='' >> used for defining the graph type >> (default=scatter)
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='reg')
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='kde')
```
