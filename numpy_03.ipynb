```python
import numpy as np
```

```python
array_1 = np.random.randint(1, 3, (3, 3))
array_2 = np.random.randint(1, 3, (3, 3))

print(f"Array 1:\n{array_1}\n{"**"*12}")
print(f"Array 2:\n{array_2}")
```

```python
# flatten() >> return a 1D array from the original array of any dimention
array_1.flatten()
```

```python
# the size of both the array remains the same
print(f"Original array size: {array_1.size}")
print(f"Flatten array size: {array_1.flatten().size}")
```

```python
# expand dimention
array_3 = np.array([1, 2, 3, 4])
print(f"Array:\n{array_3}\nDimention: {array_3.ndim}D")
```

```python
# increase the dimention by 1 in the provided axis
# 0 -> x axis
# 1 -> y axis
np.expand_dims(array_3, axis=0) # expanded on: x-axis
```

```python
np.expand_dims(array_3, axis=1) # expanded on: y-axis
```

```python
print(f"Array:\n{array_2}\n")
array_expand = np.expand_dims(array_2, axis=0)
print(f"Expended array:\n{array_expand}")
```

```python
array_a = np.array([[1], [2], [3]])
print(f"Array:\n{array_a}\nDimention: {array_a.ndim}D")
```

```python
# sqweezing the dimention
np.squeeze(array_a) # remove/lower 1 dimention from the original array
```

```python
# repeating the elements of an array
np.repeat(array_a, 4)
```

```python
print(f"Original array:\n{array_1}")
array_a_repeated = np.repeat(array_1, 2)
print(f"Repeated array: {array_a_repeated}")
```

```python
# repeating elements in rows
np.repeat(array_1, 2, axis=0)
```

```python
# repeating elements in columns
np.repeat(array_1, 3, axis=1)
```
