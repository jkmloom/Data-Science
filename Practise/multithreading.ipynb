```python
import time
import threading # pre-installed python module
start = time.perf_counter()

def test_func():
    print("do something")
    print("sleep for 1 second")
    time.sleep(1)
    print("done with sleeping")

# run the program on two threads
t1 = threading.Thread(target = test_func)
t2 = threading.Thread(target = test_func)

t1.start() # to start the thread execution
t2.start()

t1.join() # join-> first execute t1 and t2 threads before the main thread
t2.join()

end = time.perf_counter()
print(f"The program executed in {round(end-start, 2)}")
```

```python
import time
import threading # pre-installed python module
start = time.perf_counter()

def test_func():
    print("sleep for 1 second")
    time.sleep(1)
    print("done with sleeping")

threads = []
for i in range(10):
    t = threading.Thread(target = test_func)
    t.start()
    threads.append(t)

for thread in threads:
    thread.join()

end = time.perf_counter()
print(f"The program executed in {round(end-start, 2)}")
```

- since test_func() is called 10 times it should have taken ~10 secs
- but due to multithreading it executed in ~1 sec

```python
# using multithtreading with function that takes the argument

import time
import threading # pre-installed python module
start = time.perf_counter()

def test_func(args):
    print(f"sleep for {args} second")
    time.sleep(args)
    print("done with sleeping")

threads = []
for i in range(5):
    t = threading.Thread(target = test_func, args = [2])
    t.start()
    threads.append(t)

for thread in threads:
    thread.join()

end = time.perf_counter()
print(f"The program executed in {round(end-start, 2)}")
```

```python
# use case
# multithreading works well with i/o bound tasks
# meaning where someone has to wait for the outputs to be produced
# examples: reading-writing the files, network communication, database queries, etc.
```

```python
# each url below contain some text file -> we'll be downloading it using multithreading
import time
import threading

start = time.perf_counter()
url_list = [
    "https://github.com/itsfoss/text-files",
    "https://github.com/garetharnold/cultdeadcow"
]

data_list = ['data1.txt', 'data2.txt']

import urllib.request

def file_download(url, filename):
    urllib.request.urlretrieve(url, filename)

threads = []
for i in range(len(url_list)):
    t = threading.Thread(target = file_download, args = (url_list[i], data_list[i]))
    t.start()
    threads.append(t)

for thread in threads:
    thread.join()

end = time.perf_counter()
print(f"Execution time: {round(end-start, 2)} seconds") 
```

```python
# multithreading (using concurrent.features) >> this keeps code concise

import time
import concurrent.futures
import urllib.request

start = time.perf_counter()
url_list = [
    "https://github.com/itsfoss/text-files",
    "https://github.com/garetharnold/cultdeadcow"
]

data_list = ["data1_code2.txt", "data2_code2.txt"]

def file_download(url, filename):
    urllib.request.urlretrieve(url, filename)

    with concurrent.futures.ThreadPoolExecutor() as executor:
        executor.mp(file_download, url_list, data_list) # args >> function name with arguments of the functions

end = time.perf_counter()
print(f"The program executed in {round(end-start, 2)} seconds.")
```
