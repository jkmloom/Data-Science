```python
# accessing elemets of an array
print(f"Array: {arr}")
print(f"0th element of array: {arr[0]}")
```

```python
# array is mutable
print(f"Array (before): {arr}")
arr[0] = 5
arr[-1] = "@jkmloom"
print(f"Array (after): {arr}")
```

```python
ar = [1, 2, 3, 4, 5]
a = ar # 'a' and 'ar' will point to same memeory location
# can on any will be reflecte on both
print(f"a: {a}")
print(f"ar: {ar}")
```

```python
# shallow copy >> chane in one array will leadto change in the other array
a[0] = 50
print(f"a (after): {a}")
print(f"ar (after): {ar}")
```

```python
# Deep Copy >> different memory location for similar data
b = a.copy()
print(f"Array a: {a}\nArray b: {b}")
```

```python
b[0] = 55
print(f"Array a: {a}\nArray b: {b}")
```

```python
# Multiple approches to generate array

# Construct an array be executing a function over each coordinate
arr1 = np.fromfunction(lambda i, j: i==j, (3, 3))
print(arr1)
```

```python
print(f"Array Dimention (arr1): {arr1.ndim}D")
print(f"Shape of Array (arr1): {arr1.shape}")
```

```python
arr1.size # returns number of elements in an array
```
