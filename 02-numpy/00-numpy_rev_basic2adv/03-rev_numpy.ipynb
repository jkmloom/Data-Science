```python
import numpy as np
import matplotlib.pyplot as plt
```

working on real datasets via numpy

```python
# data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000], # pizza hub
    [2, 120000, 130000, 160000, 140000], # burger point
    [4, 130000, 160000, 200000, 234510], # beijing bites
    [5, 120000, 230000, 140000, 150000]  # paradise biryani
])
```

```python
print(f"{'===='*9}\n==== restro sales_data analysis ====\n{'===='*9}\n")
print(f"sales data shape -> {sales_data.shape}\n{'-----'*9}") # dataset shape
print(f"sample data for 1st 3 restro ->\n{sales_data[:3]}") # first 3 rows (via numpy)
```

```python
sales_data
```

```python
# total sales per year
print(f"total each year -> {np.sum(sales_data[:, 1:], axis=0)}")
# axis=0 >> columns
```

```python
# minimum sales per restaurant
min_sales = np.min(sales_data[:, 1:], axis=1)
# axis=1 >> rows
print(f"minimum sales :: {min_sales}")
```

```python
# maximum sales per year
max_sales = np.max(sales_data[:, 1:], axis=0)
# axis=0 >> columns
print(f"maximum sales each year :: {max_sales}")
```

```python
# average sales per restaurant
mean_sales = np.mean(sales_data[:, 1:], axis=1)
avg_sales = np.average(sales_data[:, 1:], axis=1)
# np.mean and np.average are technically same

print(f"average sales per year (np.mean()) :: {mean_sales}")
print(f"average sales per year (np.average()) :: {avg_sales}")
```

```python
# cumelative sum
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(f"cumelative sales ::\n{cumsum}")
```

```python
# just cause i was curious >> ignore
plt.Figure(figsize=(10, 6))
plt.plot(np.average(cumsum, axis=0))
plt.title('average cumelative sum accross all restaurants')
plt.xlabel('years')
plt.ylabel('sales amount')
plt.grid(True)
plt.show()
```

```python
vector_1 = np.array([1, 2, 3, 4, 5])
vector_2 = np.array([6, 7, 8, 9, 10])

# vector addition
print(f"vectors addition: {vector_1 + vector_2}")
```

```python
# vector multiplication
print(f"vector multiplication: {vector_1 * vector_2}")
```

```python
# dot-product
print(f"dot product: {np.dot(vector_1, vector_2)}")
```

```python
# calculating angle -> (understanding may need a bit of reading online)
angle = np.arccos(np.dot(vector_1, vector_2) / (np.linalg.norm(vector_1) * np.linalg.norm(vector_2)))
print(f"angle: {angle}")
```

```python
# vectorised operation
restaurant_types = np.array(['biryani', 'chinese', 'pizza', 'burger', 'cafe'])
vectorized_upper = np.vectorize(str.upper)

print(f"original (restaurant_type): {restaurant_types}")
print(f"vectorized upper:           {vectorized_upper(restaurant_types)}")
```

```python
# broadcasting -> below is an example
monthly_avg = sales_data[:, 1:] / 12
print(f"broadcasting example:\n{monthly_avg}")
```
