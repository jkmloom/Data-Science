> **Question 1: What is the difference between descriptive statistics and inferential statistics? Explain with examples.**

Descriptive Statistics: Focuses on summarizing and organizing data so it can be easily understood. It describes the characteristics of a known dataset without making generalizations beyond that specific data.

Example: Calculating the average test score of students in a specific classroom (e.g., "The average math score in Class A is 85%").

Inferential Statistics: Uses a sample of data to make inferences, predictions, or generalizations about a larger population. It involves probability to determine how confident we can be that the sample results apply to the entire population.

Example: Surveying 500 voters to predict the outcome of a national election containing millions of voters.

> **Question 2: What is sampling in statistics? Explain the differences between random and stratified sampling.**  

Sampling: The process of selecting a subset of individuals or observations from a larger population to estimate the characteristics of the entire population.

Random Sampling: Every individual in the population has an equal and independent chance of being selected. It is unbiased but might accidentally underrepresent specific minority subgroups by pure chance.

Stratified Sampling: The population is first divided into distinct, non-overlapping subgroups (called strata) based on a specific characteristic (e.g., age, gender, income). Then, random sampling is performed within each stratum to ensure that all subgroups are proportionally represented in the final sample.

> **Question 3: Define mean, median, and mode. Explain why these measures of central tendency are important.**  

**Mean:** The arithmetic average of a dataset, calculated by summing all values and dividing by the total number of values ![mean](./mean.png)  

**Median:** The exact middle value of a dataset when it is ordered from least to greatest. If there is an even number of observations, it is the average of the two middle numbers.

**Mode:** The value that appears most frequently in a dataset. A dataset can have one mode, multiple modes, or no mode at all.

Importance: These measures provide a single summary value that represents the center or typical value of a distribution. They help quickly understand the general behavior of the data and serve as a baseline for more complex statistical analysis.

> **Question 4: Explain skewness and kurtosis. What does a positive skew imply about the data?**  

Skewness: Measures the asymmetry of a data distribution around its mean. A normal distribution has zero skewness (it is perfectly symmetrical).

Kurtosis: Measures the "tailedness" of a distribution relative to a normal distribution. High kurtosis indicates heavy tails (more extreme outliers), while low kurtosis indicates light tails (fewer outliers).

Positive Skew Implication: A positive skew (or right-skewed distribution) implies that the tail on the right side of the distribution is longer or fatter. The bulk of the data values are clustered on the left (lower values), and the mean is typically greater than the median because it is pulled higher by a few extreme high values.

> **Question 5: Implement a Python program to compute the mean, median, and mode of a given list of numbers.**  
```numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]```  

````python
import statistics

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

mean_val = statistics.mean(numbers)
median_val = statistics.median(numbers)
# multimode is used in case there are multiple modes (which is true here: 12, 19, 24)
mode_val = statistics.multimode(numbers)

print(f"Mean: {mean_val}")
print(f"Median: {median_val}")
print(f"Mode(s): {mode_val}")
````

> **Question 6: Compute the covariance and correlation coefficient between the following two datasets provided as lists in Python:**  

`list_x = [10, 20, 30, 40, 50]`  

`list_y = [15, 25, 35, 45, 60]`

````python
import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

# Compute covariance matrix and extract the covariance between x and y
covariance = np.cov(list_x, list_y)[0][1]

# Compute correlation coefficient matrix and extract the correlation between x and y
correlation = np.corrcoef(list_x, list_y)[0][1]

print(f"Covariance: {covariance}")
print(f"Correlation Coefficient: {correlation:.4f}")
````

> **Question 7: Write a Python script to draw a boxplot for the following numeric list and identify its outliers. Explain the result:**  
`data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]`

````python
import matplotlib.pyplot as plt
import numpy as np

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

# Identify Outliers using IQR method
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [x for x in data if x < lower_bound or x > upper_bound]
print(f"Outliers: {outliers}")

# Draw Boxplot
plt.boxplot(data, vert=False)
plt.title("Boxplot of Data")
plt.xlabel("Values")
plt.show()
````

> **Question 8: You are working as a data analyst in an e-commerce company. The marketing team wants to know if there is a relationship between advertising spend and daily sales.**
**Explain how you would use covariance and correlation to explore this relationship.**  
**Write Python code to compute the correlation between the two lists:**  
`advertising_spend = [200, 250, 300, 400, 500]`  
`daily_sales = [2200, 2450, 2750, 3200, 4000]`

Explanation: I would use covariance to determine the direction of the relationship between advertising spend and daily sales. A positive covariance would indicate that as ad spend increases, daily sales tend to increase as well. However, because covariance is sensitive to the units of measurement (dollars vs. units sold), I would use the correlation coefficient to determine both the direction and the strength of the linear relationship on a standardized scale from -1 to +1. A correlation close to +1 strongly implies that higher ad spend is highly associated with higher sales.

````python
import numpy as np

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

correlation = np.corrcoef(advertising_spend, daily_sales)[0][1]
print(f"Correlation between Ad Spend and Daily Sales: {correlation:.4f}")
````

> **Question 9: Your team has collected customer satisfaction survey data on a scale of 1-10 and wants to understand its distribution before launching a new product.**  
**Explain which summary statistics and visualizations (e.g. mean, standard deviation, histogram) you'd use.**  
**Write Python code to create a histogram using Matplotlib for the survey data:**  
`survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]`  

Explaination:  
Summary Statistics: I would use the mean and median to identify the average or typical customer satisfaction level. The standard deviation is crucial to understand the spread or consensus; a low standard deviation means most customers agree (scores are tightly clustered), while a high standard deviation indicates polarizing opinions.

Visualizations: A histogram is the best choice here because it visualizes the frequency distribution of the scores. It easily reveals the overall shape (e.g., normal, skewed right if people are mostly unhappy, or skewed left if mostly happy), indicating the most common score brackets at a glance.

````python
import matplotlib.pyplot as plt

survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

plt.hist(survey_scores, bins=range(4, 12), edgecolor='black', alpha=0.7)
plt.title("Customer Satisfaction Survey Distribution")
plt.xlabel("Satisfaction Score (1-10)")
plt.ylabel("Frequency (Number of Customers)")
plt.xticks(range(1, 11))
plt.grid(axis='y', alpha=0.75)
plt.show()
````
