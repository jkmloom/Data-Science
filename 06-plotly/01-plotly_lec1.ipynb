```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
# generally used via 2 modules:
# 1. graph_object, and
# 2. express

import plotly.graph_objects as go
import plotly.express as px
```

## using graph_object
here we're using it as **go** > import plotly.graph_object as go

```python
# creating a figure 'fig'
fig = go.Figure()
# adding data to the figure
fig.add_trace(go.Scatter(x=[1, 2, 3, 4, 5], y=[1, 7, 8, 2, 6], mode='markers'))
# var_name.add_trace(go.Scatter()) : go.Scatter() >> graph type specified
# go.Scatter(mode='marker') : mode='markers' >> specified data representation
```

```python
# creating line chart using plotly
fig = go.Figure()
fig.add_trace(go.Scatter(x=[1, 2, 3, 4, 5], y=[1, 7, 8, 2, 6], mode='lines'))
# moe='lines' >> for line graph (on scatter graph)
```

```python
# loading a dataset from seaborn
tips = sns.load_dataset('tips')
tips.info()
```

```python
# creating a scatter plot of the loaded dataset: tips
fig = go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill, y=tips.tip, mode='markers'))
```

```python
# creating a line chart of the above data
fig = go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill, y=tips.tip, mode='lines'))
# practically this'll make no sense (line chart/graph is not sutable for this data)
```

```python
# creating a bar chart
fig = go.Figure()
fig.add_trace(go.Bar(x=[1, 2, 3, 4, 5], y=[6, 7, 8, 9, 2]))
# var_name.add_trace(go.Bar()) : go.Bar() >> used for bar graph, and no mode used here
```

```python
# creating an histogram
fig = go.Figure()
fig.add_trace(go.Histogram(x=tips.total_bill))
# use go.Histogram for creating a histogram
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill, y=tips.tip, mode='markers'))
```

```python
# custom marker size based of data/parameter
fig = go.Figure()
fig.add_trace(go.Scatter(x=tips.total_bill, y=tips.tip, mode='markers', marker_size=tips.total_bill))
# here based on 'tip' column
```

```python
# checking datatypes used
tips.dtypes
```

```python
# creating a 3d plot
fig = go.Figure()
fig.add_trace(go.Scatter3d(x=tips.total_bill, y=tips.tip, z=tips.tip, mode='markers', marker_size=10*tips['tip']))
# use : go.Scatter3d(x, y, z)
```

```python
# creating a 3d line chart
fig = go.Figure()
fig.add_trace(go.Scatter3d(x=[1, 2, 3, 4, 5], y=[3, 8, 5, 2, 6], z=[1, 2, 7, 3, 5]))
```

## using express
we're using it as **px** >> import plotly.express as px

```python
# custom data
x = [1, 2, 3, 4]
y = [10, 13, 15, 11]
```

```python
# creating a simple scatter plot using plotly.express
fig = px.scatter(x, y)
fig.show()
```

```python
fig = px.line(x, y)
fig.show()
```

```python
# another custom data
x = ['A', 'B', 'C', 'D']
y = [10, 5, 6, 14]
```

```python
# creating bar plot
fig = px.bar(x=x, y=y)
fig.show()
```

```python
# another custom data
x = [5, 10, 3, 4, 5, 5, 6, 6, 10, 10, 11, 12, 3, 4]
```

```python
# creating histogram
fig = px.histogram(x)
fig.show()
```

```python
# another custom data
x = ['A', 'B', 'C']
y = [5, 10, 3, 25, 12, 33, 100]
```

```python
# creating box-plot
fig = px.box(x, y)
fig.show()
```

```python
# creating heat-map
# custom data
data = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
# heat-map
fig = px.imshow(data)
fig.show()
```

```python
# creating a bubble chart

# custom data
x = [1, 2, 3, 4]
y = [12, 34, 14, 15]
# bubble-chart
fig = px.scatter(x, y, size=[10, 30, 50, 100])
fig.show()
```

```python
# custom data
x = [1, 2, 3, 4]
y = [8, 5, 12, 3]
z = [3, 4, 5, 6]

# 3d bubble chart
fig = px.scatter_3d(x=x, y=y, z=z)
fig.show()
```

```python
# creating a contour plot

# custom data
x = [1, 2, 3, 4, 3, 23, 80]
y = [5, 8, 12, 6, 12, 35, 60]

# contour plot
fig = px.density_contour(x, y)
fig.show()
```
