# **probablity distribution function with CDF**
```text
CDF --> cummulative distributive function
```

## PMF --> probablity mass function  
- discrete random variable  
- example, rolling a dice --> possible outcome {1, 2, 3, 4, 5, 6}  
- probablity for each possible value is same for one dice --> 1/6  
- when all the possible outcomes have same probablity the we call it **uniform distribution**.  

````python
import matplotlib.pyplot as plt

possible_outcomes = [1, 2, 3, 4, 5, 6]
probablities = [1/6, 1/6, 1/6, 1/6, 1/6, 1/6]

plt.bar(possible_outcomes, probablities)
plt.title("rolling a dice")
plt.xlabel("possible outcomes")
plt.ylabel("probablities")
plt.show()
````

## CDF --> cummulative distributive function

**cummulative distributive function (CDF) for the above function -->**  
P(x<=1) = 1/6  
P(x<=2) => P(x=1) + P(x=2) = 1/6 + 1/6 = 2/6 or 1/3  
P(x<>=3) => P(x=1) + P(x=2) + P(x=3) = 3/6 or 1/2
ans so on..  
```text
at last we'll get 6/6 = 1
```

````python
# bar graph representing the CDF data for above function

import matplotlib.pyplot as plt

print("possible outcomes", possible_outcomes)
cdf_outcomes = [1/6, 2/6, 3/6, 4/6, 5/6, 6/6]

plt.bar(possible_outcomes, cdf_outcomes)
````

## PDF --> probablity density function  
- random variable is continuous in nature.

````python
students_height = [140, 150, 160, 170, 180, 190, 200]
occurance_probablity = [0.01, 0.02, 0.03, 0.04, 0.03, 0.02, 0.01]

plt.plot(students_height, occurance_probablity, marker='o', linestyle='--', color='purple')
plt.bar(students_height, occurance_probablity, color="purple")
plt.grid(linestyle='--')
plt.title('probablity graph for students height')
plt.xlabel('students height')
plt.ylabel('probablity')
plt.show()
````
