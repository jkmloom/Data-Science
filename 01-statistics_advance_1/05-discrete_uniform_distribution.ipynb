# discrete uniform distribution 

> **uniform distribution** --> it reefres to a type of probablity distribution in which outcomes are equally likely occure.

- probablity distribution is of 2 types:
1. probablity mass function, and  
2. probablity density function  

**uniform distribution types:**  
1. discrete uniform distribution, and  
2. continuous uniform distribution

## discrete uniform distribution
> here the outcomes are discrete and have same probablity.  

*examples:*  
1. rolling a dice --> 1/6 for every possible outcome  
2. tossing a coin --> 1/2 for head and same for the tail  
3. picking up a card from a well shuffled deck --> 1/52 for all cards  

```python
# all the 3 examples mentioned above are uniform distribution
```

**notation of uniform distribution:** `u(a, b)` or `uniform{a, b}`

`probablity mass function (PMF) = 1/n`  
where n = 6 --> total number of possible oucomes after rolling a dice  
generalized formulation for n --> `n = b - a + 1`  
putting in the values to the general formual of n --> 6 - 1 + 1 = 6

````python
import matplotlib.pyplot as plt

possible_outcomes = [1, 2, 3, 4, 5, 6]
probablities = [1/6, 1/6, 1/6, 1/6, 1/6, 1/6]

plt.bar(possible_outcomes, probablities)
plt.title("probablity mass function (PMF) of discrete uniform distribution")
plt.xlabel("possible outcomes")
plt.ylabel("probablities")
plt.show()
````

````python
print("possible outcomes", possible_outcomes)
cdf_outcomes = [1/6, 2/6, 3/6, 4/6, 5/6, 6/6]

plt.bar(possible_outcomes, cdf_outcomes)
plt.title('cumulative distribution function of discrete uniform distribution')
plt.xlabel('possible outcomes')
plt.ylabel('probablity')
````

**mean of discrete uniform distribution**  --> `(a + b) / 2`  
**variance of discrete uniform distribution** --> `(n^2 - 1) / 12`

***::  example  ::***  
rolling a dice --> possible outcomes --> 1, 2, 3, 4, 5, 6  
mean = (1 + 2 + 3 + 4 + 5 + 6) / 6 = 3.5

> **expected value  -->**  average value of repetations of experiment in a long run  

or we can also say that -> *long term average value of a random variable* --> expected value

- **mean** is used for frequency distribution, and  
- **expected** value is used for probablity distribution

`variance(x) = expected(x^2) - (expected(x))^2`
