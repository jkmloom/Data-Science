# **types of distribution**

mainly there are 2 types :-  
1. probablity density function -->  
    - normal distribution / bell shaped distribution / gaussion distribution  
    - standard normal distribution  
    - log normal distribution  
    - chi-square distribution  
    - F-distribution  
2. probbality mass function
    - bernoulli distribution  
    - binomial distribution  
    - poission distribution  

- uniform distribution (it's of both the types)  
    - discrete uniform distribution, and  
    - continuous uniform distribution
