# **probablity distribution function**

**randome variable >>** a set of possible values from a random experiment.  
- value for a random value is unknown.  
- a function that assigns values to each o fexperiment's outcome.

example --> `tossing a coin` --> {H, T}  
we can map the possible set of outcomes to `x = {0, 1}`  
but, without it --> **P(x = H) = 1/2** and **P(x = T) = 1/2**

can we write a general function for the above example ? --> yes!  
we may say, `1/n` (where **n** is total number of outcome)  

example --> tossing a coin (outcomes -> H, T)  
P(H) = 1/2  

example --> rolling a dice (outcomes -> 1, 2, 3, 4, 5, 6)  
P(1 or 2 or 3 or 4 or 5 or 6) = 1/6  
**general formula -->** 1/n, where n = 6  
```text
    1/n is a function used to find probablity
```

## functions type (based on outcomes of experiment)

functions' type -->  

> **PORBABLITY MASS FUNCTION** --> outcomes are discrete values  
examples: tossing a coin, and rolling a dice  

> **PROBABLITY DENSITY FUNCTION** --> outcmoes are continuous values  
example: height of a student [160.0 cms, 160.1 cms, 160.2 cms, ..]

```text
probablity distribution function --> irrespective of outcome nature, draw outcomes in a form of distribution.

example --> rolling a dice. possible outcomes --> {1, 2, 3, 4, 5, 6}
say, P => probablity
P(1) = 1/6
P(2) = 1/6
P(3) = 1/6
P(4) = 1/6
P(5) = 1/6
P(6) = 1/6
```

````python
import matplotlib.pyplot as plt

possible_outcomes = [1, 2, 3, 4, 5, 6]
probablities = [1/6, 1/6, 1/6, 1/6, 1/6, 1/6]

plt.bar(possible_outcomes, probablities)
plt.title('probablities after rolling a dice')
plt.xlabel('possible outcomes')
plt.ylabel('probablities')
plt.show()
````

```text
since the above data is discrete, therefore its probablity mass function.
```

> another example

height of students --> [150, 160, 160, 140, 180]

````python
height = [150, 160, 160, 140, 180]
student_count = [5, 10, 12, 7, 15]

plt.bar(student_count, height)
plt.grid()
plt.title("students' height graph")
plt.xlabel("students' count")
plt.ylabel("height")
plt.show()
````
