# **measure of symmetricity & skewness**
## symmetricity
it talks about the shape of the data (aka data population)

![histogram distribution graph](./img_data/histogram_ditribution_graph.png)

considering the above graph, **tip of the distribution is the mode value**

> if anything is exactly from the sides (left and right) >> we call it symmetry.  

**example >>** in the above graph the left and right halfs of the longest bar are symmetrical.  
we can also say that the data (data population) is 50% on both the sides

## skewness
> to measure this symmetry we use the concept called >> **skewness**  

- we do measure of symmetricity  

skewness means >> inclined to some side  
- no skewness means >> nothing inclined [symmetric]  
```no inclination = symmetric value```

### skewness in data (data population)

> - **positive / right skewed data** -> data tail is on right side of the data distribution  

most of the data in the distribution lies on the right side of the graph  
**example >>** log-normal distribution  
relatinoship >> ```mean >= median >= mode```  
![relationship b/w mean, median, and mode for skewness](./img_data/skewness_relationship_mea_med_mod.png)

> - **negative / left skewed data** -> data tail is on left side of the data distribution  

most of the data lies on the left side of the graph  
relationship >> ```mode >= median >= mean```

- **if skewness =  0** >> given distribution is symmetric  
- **if -1 < skewness < 1** >> given distribution is moderately/mildly skewed
- **if skewness is >1 or <-1** >> highly skewed data  

to know if data is skewed or not >>  
1. visualization (distribution plot)  
2. skewness (calculate)  

> some of machine learning algos requires symetric data to build a model  

>if not symmetric >> some data transformation is required

#### data transformation type  
1. log transformation,  
2. exponential transformation,  
3. reciprocal transformation,  
4. square transformation,

2. **presence of outlier in the data population**
