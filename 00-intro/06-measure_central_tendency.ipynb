### >> **measures of central tendency**

**analogy for understanding >>**
- representative of a state is a ***chief minister***
- so, we can say that ***the chief minister acts as the center*** for the state (not a grammer friendly statement, but you get it)

**more analogies >>**

[1  2  3  4  5] --> could be 3

```central tendency represents the center point of a dataset.```  

this is done using:  
1. mean,  
2. median, and  
3. mode

[these are also used a lot in **exploratory data analysis, data preparation, feature engineering**, etc.]

### [| **1. mean** |]
---
say there's a population: 1, 2, 3, 4, 5  
so mean of the population is as follows:  
<p align="center">
    <img src="./img_data/mean_formulation.png">

<!-- ![mean formulation](./img_data/mean_formulation.png) -->

say x = [1, 2, 3, 4, 5]

````python
data_population = [1, 2, 3, 4, 5]
summition = sum(data_population)
n = len(data_population)

mean = summition/n
print(f"mean for the given data population: {data_population} is --> {mean}")
````

```mean has other names >> average, and arithmetic mid value of data```

**mean >>** summing up all the observations, then divided by the number of observations

### [| **2. median** |]
---
for making it easy, say >> physical mid-point of data
<p align="center">
    <img src="./img_data/median_formulations.png">
</p>

**given data >>** [4, 5, 2, 3, 1, 2]  
- sort the data >> 1, 2, 2, 3, 4, 5
- count the number of elements >> 6

**if the count is even**  
**median** = average of two middle most elements  
here in our case, 1, 2, ***2, 3***, 4, 5  
(2 + 3)/2 = 2.5

**another example >>**  
> given datset: [4, 5, 2, 3, 1]  
- sort it >> 1, 2, 3, 4, 5  
- count >> 5  
- **if count is odd >>** median is the middle most element

```physical mid point``` of the data, ```(middle most element)```  

[   1   2   3   4   5   ]  
assume the numers are human being, who's the middle most human?  
ans.>> 3

#### **scenario 1**
**case 1** >> mean = (1 + 2 + 3 + 4 + 5) / 5 = 3  
**case 2** >> mean = (1 + 2 + 3 + 4 + 1000) / 5 = 202

#### **scenario 2**
**case 1** >> median of 1, 2, 3, 4, 5 = 3  
**case 2** >> median of 1, 2, 3, 4, 1000 = 3

```the mean is affected by the outliers' value whereas median is not affected by the actual value of the outliers```

### [| **3. mode** |]
---
maximum frequency of an element in the data population

[2, 3, 1, 1, 4, 4, 4, 3, 4, 2] >> which element is repeated higest number of times?  
**ans.>> 4**

> in all of these cases (mean, median, and mode) we're trying to represent the whole data with a central value

### **when to calculate mean, median and mode? and on which type of data?**
> <i.> if given population is numerical (and continuous) data then, we calculate mean and median  
  
> <ii.> if the given population is categorical then, we calculate mode

### **use cases of central tendency >>**

| Age | Gender | Weight | Salary (in k) |
| :--- | :--- | :--- | :--- |
| 25 | m | 80 | -- |
| 26 | -- | 70 | 50 |
| -- | m | 30 | 60 |
| 23 | m | -- | 70 |
| 25 | f | 1000 | 45 |


> **missing value imputation**  
- say **age** is a continuous variable
- then, we impute / replace the missing / null value with mean if there is no outlier
- {25 + 26 + 23 + 25}/4 = 24.75 ~ 25  
.
- **gender** is a categorical value / column
- we need to find the mode (highest frequency)
- m, m, m, f -> so we assume it to be m  
.
- **weight** is a continuous variable
- we can see that there is an outlier [1000]
- so, we calculate the median value for imputation
- 80, 70, 30, 1000 -> (80 + 70) / 2 = 75  
.
- **salary** is a coninuous value
- there is no outlier value
- so, we calculate mean
- (50 + 60 + 70 + 45) / 4 = 56.25 ~56
