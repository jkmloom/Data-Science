# **11. implementation of symmetricity & skewness**

```python
data = [2, 3, 4, 5, 8, 9, 12, 40, 41, 15]
print(data)
```

```python
# measure of dispersion/spread

# 1. range
print("max of data >> ", max(data))
print("min of data >> ", min(data))
print("difference in max and min of data >>", max(data) - min(data))
```

```python
# 2. percentile
import numpy as np
np.percentile(data, [0, 25, 50, 75, 100])
```

```python
# 3. iqr
14.25 - 4.25 # quatile 3 - quatile 1
```

```python
# 4. box-plot
import seaborn as sns
sns.boxplot(data)

# upper fence >> q3+1.5*iqr
# lower fence >> q1-1.5*iqr
```

```python
# 5. variance
np.var(data)
```

```python
# 6. standard deviation
np.std(data)
```

```python
# another way for calculating variance
import statistics
statistics.variance(data) # sample variance >> denomator is n-1
```

```python
statistics.pvariance(data) # population variance >> same as np.var()
```

```python
# measure of symmetricity, skewness

# skewness
from scipy.stats import skew
skew(data) # since skewness value is 1.23, it is moderatoly skewed data
```

```python
d = [2, 3, 5, 7, 11]
skew(d) # data is almost same range >> skewness is close to 0
```
