### **how to calculate variance?**
- calculate mean
- for each number in data, substract the mean and the number
- square of difference
- calculate the average of square of the difference

````python
# suppose
x = {1, 2, 3, 3, 4, 4}
x_dash = sum(x)/6
print(f"mean of data = {x_dash}")
````

**find mean for every value of x**  
- sum of all the values in x / number of values in x  

**differences for all values of x**  
- x - x_dash (for all the values of x)  

**square of the differences for all values of x**
- (x - x_dash)^2  

**average of squared differences for all the values of x**
- sum of all [(x - x_dash)^2] / (number of all the values of x - 1)

**as variance increases, spread also increases**  
> variance is proportional to spread
