# **12. set**

> **set >>** collection of unordered unique elements  

**example >>**
A = (1, 2, 3, 4, 5)  
B = (3, 5, 7, 9)  
*below diagram represents the above data >>*  

![sets venn-diagram](./img_data/set_venn-diagram.png)

## i. intersection
**common elemets** (elements present in the intersected are of A and B) >> *refer to the venn-diagram above*  
in above example >> ```A intersection B = {3, 5}```


## ii. union
**all distinct elements** from both the sides (all elements irrepective of it being common or not)  
in above example >> ```A union B = {1, 2, 3, 4, 5, 7, 9}```

## iii. difference
items which are present only in the 1st set  
in above example >> elements of set A-B >> ```A-B = {1, 2, 4}```  
{3, 5} were common among A and B so we removed them in the difference

## iv. subset
when all the elements of B are present in A, but all the elements of A is not necessarily present (may or may not) in B then we say that **B is a subset of A**  
**example >>**  
A = {1, 2, 4, 5, 7}  
B = {2, 4, 5}  
if we say that B is the subset of A then >> all elements of B are present in A  
in this case ```true``` :: B indeed is a subset of A

## v. superset
if A is containing all elements of B, then A is superset of B  
**example >>**  
A = {1, 2, 4, 5, 7}  
B = {2, 4, 5}  
in the above example ```A is a superset of B```  
but ```B is not a supetset of A```


## vi. symmetric difference
in simple words >> opposite of intersection  
- the elements that are distinct in both sets excluding intersection  

**for example >>**  
A = {1, 2, 3, 4, 5, 6, 7}  
B = {5, 6, 7, 8, 9, 10, 11}  
for the above example teh symmetric difference will be >> ```{1, 2, 3, 4, 8, 9, 10, 11}```  
common elements were here are ```{5, 6, 7}```

- another example  

A = {4, 5, 2, 8}  
B = {2, 8, 6, 7}  
**symmetric difference >>** `{4, 5, 6, 7}`
