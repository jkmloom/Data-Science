**types >>** 1. descriptive statistics, and 2. inferential statistics
## **1. descriptive statistics**
- descriptive comes from the word 'describe'
- it involves organizing and summerizing complete data/population

**example (question/problem)>>** how good is jkmloom as at playing valo?  
- **answering this >>** performace in all the matches played, scored throughout every match, frequency and duration of gameplay, performance analytics, etc.

## **2. inferential statistics**
- comes from the word 'inference'
- one can not exectly count/study some values (example: all the trees on our planet, earth)
- it involves using sample/dedicated data that has been measured to form conclusions about the popluaiton/data.

**with given sample data we conclude something about the populaiton**  
  
**::  _ inferential statistics example _  ::**  
consider india's population (as of 2026) >> sample data (estimated) >> calculating average weigth >> considered true for whole population

**another example**  
suppose a class E5 has 150 students. we weight 70 studnets (that's more than 50%) >> we calculate the average (say 47kgs) >> we say the students of class E5 weight around 47kgs.  

___
**making it simpler**
```
exact values (when accuracy is needed) >> descriptive statistics  
when a large population/dataset is given for statistical analysis and we can't go though exact values >> inferential statistics
```

we generally have got a few techniques we use for **descriptive analysis**:
1. measure of central tendency (mean, median, mode)  
2. measure of symmetricity (skewness, kurtosis)
3. measures of dispersion (standard deviation, variance)

**inferential analysis**:  
(used for large population and when we've got time & resource constraints)  
```[population]``` >> ```[sample]``` >> ```[inferential statistical analysis]``` >> ```[anounced true for the population]```

**example >>** find the number of trees in a national park. [given: 10% of area has 2,000 trees]  
we consider a sample area out of the actual land area  
since, 10% area has 2,000 trees. total number of trees = 20,000 (from 10*2,000)
