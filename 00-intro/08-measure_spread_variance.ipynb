### **1. range**
difference between maximum and minimum value  
**example 1 >>** {1, 2, 3, 4, 5} --> range => 5 - 1 = 4  
**example 2 >>** {1, 2, 3, 4, 1000} --> range => 1000 - 1 = 999  
```outlier affects range, so we can't use range in every case to understand the spread```

### **2.2 percentile**
value below which a certain percentage of observation lies  
example >> say you got 90 percentile in an exam >> that says this many students have scored less than you   
- there is no single universally accepted universal symbol for "percentile," but it is typically denoted by the capital letter P or lowercase p accompanied by the percentile rank (e.g., P₉₀ for the 90th percentile).

<p align="center"><img src="./img_data/percentile_formulation.png"></p>
percentile rank of a number is given by = (number of values below that number / total number of values) * 100  

what is the percentile rank of 3?
given: 1, 2, 3, 4, 4, 6, 7, 7, 8, 10  
**a/c to the given formulation >>** (number of values below 3 / total number of values) * 100  
```(2/10)*100 = 20 (or P20)```  
we can say that 20% of the values in the given population are below 3

- **problem example >>** what value exists at 75th percentie?  
```value = percentile/100 * (n+1)```  
value = 75/100 * (10+1)  
=> value = 3/4 * 11 = 33/4 = 8.25 (or 8)  

from the given population >> we can say that 8th value (or the 75th percentile) is **7**  
75% of the numbers in the given population is below 7

### **3. quartile**
> the value that divides a list of numbers into quaters
- put the numbers in order
- then, cut the numbers into 4 equal parts
- the quartiles are at the cut

**example >>** 6, 8, 5, 5, 7, 3, 9  
arrange in order >> 3, 5, 5, 6, 7, 8, 9  
cut the values into quaters >> 3 | 5 | 7 | 9  
- quater 1 >> 3
- quater 2 >> 5
- quater 3 >> 7
- quater 4 >> 9
