### **1. range**
difference between maximum and minimum value  
**example 1 >>** {1, 2, 3, 4, 5} --> range => 5 - 1 = 4  
**example 2 >>** {1, 2, 3, 4, 1000} --> range => 1000 - 1 = 999  
```outlier affects range, so we can't use range in every case to understand the spread```

### **2.2 percentile**
value below which a certain percentage of observation lies  
example >> say you got 90 percentile in an exam >> that says this many students have scored less than you   
- there is no single universally accepted universal symbol for "percentile," but it is typically denoted by the capital letter P or lowercase p accompanied by the percentile rank (e.g., P₉₀ for the 90th percentile).

<p align="center"><img src="./img_data/percentile_formulation.png"></p>
percentile rank of a number is given by = (number of values below that number / total number of values) * 100  

what is the percentile rank of 3?
given: 1, 2, 3, 4, 4, 6, 7, 7, 8, 10  
**a/c to the given formulation >>** (number of values below 3 / total number of values) * 100  
```(2/10)*100 = 20 (or P20)```  
we can say that 20% of the values in the given population are below 3

- **problem example >>** what value exists at 75th percentie?  
```value = percentile/100 * (n+1)```  
value = 75/100 * (10+1)  
=> value = 3/4 * 11 = 33/4 = 8.25 (or 8)  

from the given population >> we can say that 8th value (or the 75th percentile) is **7**  
75% of the numbers in the given population is below 7

### **3. quartile**
> the value that divides a list of numbers into quaters  

![quatile concept](./img_data/quatile_concept.png)  
- put the numbers in order
- then, cut the numbers into 4 equal parts
- the quartiles are at the cut

**example >>** 6, 8, 5, 5, 7, 3, 9  
arrange in order >> 3, 5, 5, 6, 7, 8, 9  
cut the values into quaters >> 3 | 5 | 7 | 9  
- quater 1 >> 3
- quater 2 >> 5
- quater 3 >> 7
- quater 4 >> 9

**another example >>** 1, 1, 1, 1, 2, 2, 2, 3, 3, 3, 4   
in total we've got 11 numbers/elemnts.  

**QUATILE 1: Q1**  
```if total numbers is odd: q1 = (n+1)/4 = (11+1)/4 = 3```  

**QUATILE 3: Q3**  
so the 3rd value is q1 (quatile 1) >> 1  
```and for q3 = 3(n+1)/4 = 36/4 = 9```  
so the 9th value will be q3 (quatile 3) >> 3

**QUATILE 2: Q2**  
```now for q2 = (n+1)/2 = (11+1)/2 = 6```  
so the 6th value will be q2 (quatile 2) >> 2  

**QUATILE 4: Q4**  
```finally for q4 = (simply the last value) = 11```  
thus, the last 11th value is q4 (quatile 4) >> 4

**another example >>** 1, 2, 3, 3, 4, 4 (n=6, even)  

**q1 >>**  
q1 = n/4 = 3/2 = 1.5  
```if it's 1.5 then we should calculate the average of 1st and 2nd index values```  
> (1+2)/2 = 3/2 = 1.5  

**q3 >>**
q3 = 3(6)/4 = 18/4 = 4.5  
> calculating mean/average, (3+4)/2 = 7/2 = 3.5  

**q2 >>**
q2 = (n+1)/2 = 7/2 = 3.5  
> calculating mean/average = (3+3)/2 = 3

### inter quartile range
```IQR = Q3 - Q1```  
since, IQR deals with Q3 and Q1, it is not affected by outliers >> the porblem in range is resolved here

**example >>**  
given population >> 2, 4, 4, 5, 6, 7, 8  
IQR -> 7 - 4 = 3  

#### **another example >>**  
given population >> 2, 3, 3, 3, 3, 4, 4, 5, 5, 5, 6, 6, 6, 7, 8, 99 | (n=16)  
> Q1 = 25th percentile = 25/100 * 16 = 4th value = 3  

> Q3 = 75th percentile = 75/100 * 16 = 12th value = 6  

> Q2 = 50th percentile = 5+5/2 = 5  

now, we can say the following >>  
Q0 = 2,  
Q1 = 3,  
Q2 = 5,  
Q3 = 6, and  
Q4 = 99  

now, ```IQR = Q3 - Q1 = 6 - 3 = 3```

> ***use case of IQR >>*** to detect outliers  

outliers are extreme values in a given data populition  

outlier <- |---------------------| -> outlier  

**lower fence >>** ```Q1 - 1.5 * IQR```  
**upper fence >>** ```Q3 + 1.5 * IQR```

say, Q1=3, Q3=6, and IQR=3  
> lower fence = 3 - 1.5 * 3 = -1.5  

> upper fence = 6 + 1.5 * 3 = 4.5 + 6 = 10.5
