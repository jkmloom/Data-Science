```python
# mean
age = [28, 56, 35, 28, 76, 89]
mean = sum(age)/len(age)
print(f"mean of {age} is >> {mean}")
```

```python
import numpy as np
np.mean(age)
```

```python
height = [170, 180, 100, 120, 122]
np.mean(height)
```

```python
# median
nos = [4, 2, 3, 7, 8]
np.median(nos)
```

```python
# mode
nos = [4, 2, 2, 3, 7, 8]
# numpy doesn't support mode

# so from scipy we import stats
from scipy import stats
```

```python
stats.mode(nos)
```

```python
# we can also use another module / method
import statistics
statistics.mode(nos)
```

```python
# proof >> mean is affected by outliers: height[1]
height = [170, 1800, 100, 120, 122]
np.mean(height)
```

```python
np.median(height) # median not affacted that much
```

```python
# another example
height = [160, 120, 125, 340, 150]
np.mean(height)
```

```python
np.median(height)
```

```python
# introducing a outlier
height = [1600, 120, 125, 340, 150]
print(f"mean's affected: {np.mean(height)}")
print(f"median's not affacted (or least affected): {np.median(height)}")
```
