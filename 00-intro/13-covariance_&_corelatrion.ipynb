## **covariance**
![covariance formulation](./img_data/covariance_formulation.png)  
where x, and y are two features

**co + variance**  
co --> existing together  
variance --> spread of data  

formula for variance of x >>  
![variance of x](./img_data/variance_x_formula.png)  
> **covariance** --> you're trying to understand the relationship of a feature with respect to the other feature

covariance is used to study the relationship between 2 variables.  
**covariance type >>**  
1. direct relationship covariance (aka `positive covariance`), and  
2. indirect relationship covariance (aka `negative covariance`)

**example >>**  
say, x = {2, 3, 6, 1} so x_bar ~= 3, and  
y = {3, 5, 6, 8} so y_bar ~= 5.5  

now,
`(x - x_bar)` for all values of x = {-1, 0, 3, -2}  
`(y - y_bar)` for all values of y = {-2.5, -0.5, 0.5, 2.5}  

now, sum of all the products of all the values in accordance to x's and y's divided by n-1 >>  
=> ((-1 * -2.5) + (0 * -0.5) + (3 * 0.5) + (-2 * 2.5)) / 4 - 1  
=> (2.5 + 0 + 1.5 - 0.5) / 3 = -1/3  
=> -0.33  
**hence, covariance for the above data/populaiton is -0.33** --> `negative value`  
```text
    interpretation >> the two features x and y are negatively related.
    so, here we have negative covariation (aka --> indirect relationship covariance)
```

**another example >>**  
now, from the formulation below >>  
![covariance formulation](./img_data/covariance_formulation.png)  
we can say that >>  
=> ((2 - 4)(3 - 5) + (4 - 4)(5 - 5) + (6 - 4)(7 - 5)) / 3 - 1  
=> (4 + 0 + 4) / 2 = 4  
**hence, covariance for the given data/population is 4** --> `positive value`  
```text
    interpretation >> x ad y are havnig a positive relationship.
    so, here we have positive covariance (aka --> direct relationship covariance)
```

### disadvatages of covariance
consider features x & y and A & B  
say, cov(x, y) = 50, and cov(A, B)=100  
**can i say that the relationship between x and y is half of A and B ?** --> `no!`

```text
'cause the features x, y, A, and B can have any values. meaning they can be anything between - infinity to infinity.
range of covariance is from minus (-) infniity to infinity
```
> 1. we ***can not compare the strength/magnitude*** of relationship.  
no standardized technique to measure the magnitude/strength  

> 2. covariance has dimentions.  
for a data population having the following feilds >> height (ft.), weight (kg), transaction_amount (inr.)  
then, covariance of **transaction_amount** and **height** will be in inr.ft.  
similarly, covariance of **height** and **weight** may be in ft.kg  
***say these resultent units are dimentions, we can see that they're different from one another (in this case inr.ft and ft.kg)***
```python
    # so we simple can not compare these values. hence, different dimentions of covariance can not be compared
```

### solution to the disadvantages
1. define a fixed range (maybe -1 to 1)  
2. it should be dmiention-less
