**major types of data**
1. quantitative data (numerical/measurable data)  
|___ i. discrete data  
|___ ii. continuous data  

2. quality data (categorical data)  
|___ i. nominal data  
|___ ii. ordinal data

### **1. quantitative data**
#### **i. discrete data**
**(sub division of quantitative data)**  
- it'll consist of specific, finite, whole numbers values  
example: number of children, number of mobile phones, etc.

#### **ii.continuous data**
**(sub division of quantitative data)**
- any value in range (or real numbers)  
example: duration of a movie, height of a person, distance to a place

### **2. qualitative data**
#### **i. nominal data**
**(sub division of qualitative data)**  
- no order in categorical data (naturally)  
example: colors (one can not be better than others), human race, bloog group, marital status, etc.

#### **ii. ordinal data**  
**(sub division of qualitative data)**  
- order in data (ranking or following a pattern)  
example: subject grade (student rank), game scores (player level), salary (low, mid, high), etc.


**use case >>**
- ```to do analysis or preparnig the data for modelling, knowing the data type is very important```
- ```sometimes data types are needed to be converted```
for example:
```python
    char_map = {'A': 0, 'B': 1, 'C': 2}
    chars = ['A', 'B', 'C']
    print(f"before: {chars}")
    numbers = [char_map[c] for c in chars]
    print(f"after: {numbers}")  
```

````python
# categorical data representation/convrsion in descrete numbers
char_map = {'A': 0, 'B': 1, 'C': 2}
chars = ['A', 'B', 'C']
print(f"before: {chars}")
numbers = [char_map[c] for c in chars]
print(f"after: {numbers}")  
````
