# **03. types of sampling**

1. simple random sampling (imp.)
2. stratified sampling (imp.)
3. cluster sampling
4. systematic sampling

### **i. simple random sampling**
every member of the population has an equal chance of being selected in the sample.  
**meaning** >> for a population size of 12, every member has 1/12 probablity of being selected


**disadvantage**
1. possibility of members not being part of sample from a certain group  
example: among 140cr. people in india, while taking sample (usnig srs-simple random sampling) members from smaller/least populated states might not be selected in sample.  

### **ii. stratified sampling**
strata >> layers or groups  
members are selected from everygroup of the population  
this was absent in simple random sampling  
in some cases we may want to use **stratified sampling** over **simple random sampling**

### **iii. cluster sampling**
divides the population into groups/clusters  
some of these clusters are selected at random  
all the individuals/members in the cluster are selected in the sample.

**example >>**  
covid cases in india (based of case density)  
say bihar, gujarat, up, and dehi are the part of smaple and bihar has the lowest case so we use green to represent bihar while delhi has the higest no. of cases so we color it red

### **iv. systematic sampling**
every nth element will be selected  
for example:

````python
import numpy as np
cluster_1 = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
sampling_logic = np.array(range(0, 10, 2))
# print(sampling_logic)
for i in sampling_logic:
    print(cluster_1[i], end=' ')
````

here the nth terms are every alternating value in the variable ```data```

there are many more types of sampling techniques but these are the 4 major ones we need to care about at this point.
