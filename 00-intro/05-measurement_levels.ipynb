**scales 4 major type >>**
1. **nominal scale data** -> qualitative measures (non-numeric categories),
2. **ordinal scale data** -> qualitative measures (non-numeric categories),
3. **interval scale data** -> quantitative measures (numeric measurements), and
4. **ratio scale data** -> quantitative measures (numeric measurements)

## **I. nominal scale data**
- qualitative or categorical data
- no order in the data
- example >> gender, ethnicity, color, location, etc.

**example of nominal data** >> say number of people like red=5, blue=3, yellow=2  
so, we can say that likes for red -> 50%, blue -> 30%, and for yellow -> 20%  
*this could be represented via a pie-chart or any other kind of plot*

```python
import matplotlib.pyplot as plt

colors = ['red', 'blue', 'yellow']
likes = [50, 30, 20]

plt.pie(likes, labels=colors, autopct="%1.2f%%")
plt.title('percentage of colors liked')
plt.show()
```

## **ordinal scale data**
- order and rand has a meaning/reason
- difference can not be measured

*could be represented via pie chart or any other kind of plot*  
- **example 1 >> educational qualification** -> 10th, 12th, ug, pg, phd, etc.  
- **example 2 >> salary** -> high, mid, low
- **example 3 >> performance ranking of students** -> 1st, 2nd, 3rd, 4th, 4th, ...

```python
salary_grade = ['high', 'mid', 'low']
people_num = [20, 17, 13]
plt.pie(people_num, labels=salary_grade)
plt.title("salary category population")
plt.show()
```

## III. interval scale data
- the rank and order has a meaning
- difference can be measured (exception: excluding ratio)
- doesn;t have compulsorily starting value = 0

**example 1. >> score** : 90, 85, 70, 50  
- has a pattern: scores are in decending order
- the difference has a meaning [say 90-85 = 5]

**example 2 >> height** : 137cms, 154cms, 175cms  
- pattern: ascending order
- difference has a meaning [say (175-137)cms = 38cms -> person 1 is 38cms shorter than person 3]

**another examples >>** weight, company revenue, temperature, etc.  
visual representations that coulld be used: scatter plot, histogram, line chart, etc.

## IV. raito scale data
- order and rank has a meaning
- difference and ratio are measurable
- has a compulsory starting point/value = 0

**example 1 >>** person A = 80kg, and person B = 40kg  
80/40 = 2:1 -> person A is twice the weight of person B  

|-------------------------|----------------------------|  
0<----------------------B------------------------->A

**another examples >>** height, weight, time, age, etc.
